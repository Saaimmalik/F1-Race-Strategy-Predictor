```python
race_laps = laps.copy()
```

```python
race_laps["LapTimeSeconds"] = (
    race_laps["LapTime"].dt.total_seconds()
)
```

```python
race_laps = race_laps.dropna(
    subset=["LapTimeSeconds"]
)
```

```python
race_laps = race_laps[
    race_laps["Compound"].isin(
        ["SOFT", "MEDIUM", "HARD"]
    )
].copy()
```

```python
race_laps = race_laps[
    race_laps["LapNumber"] > 1
].copy()
```

```python
lap_median = (
    race_laps
    .groupby("LapNumber")["LapTimeSeconds"]
    .median()
    .reset_index(name="MedianLapTime")
)
```
